```python
analysis = analysis.drop_duplicates()
```

```python
analysis = analysis.dropna()
```

```python
analysis["Total"] = analysis["Unit price"] * analysis["Quantity"]
```

```python
analysis.groupby("Product line")["Quantity"].sum().sort_values(ascending=False).head(5)
```

```python


analysis.groupby("Product line")["Sales"].sum().sort_values(ascending=False)
```

```python
analysis.groupby("Branch")["Sales"].sum().sort_values(ascending=False)
```

```python
analysis.groupby("Branch")["Sales"].mean()
```

```python
analysis["Payment"].value_counts()
```

```python
summary = analysis.groupby("Branch").agg(
    Total_Revenue=("Sales", "sum"),
    Total_Transactions=("Invoice ID", "count"),
    Average_Basket=("Sales", "mean")
)

summary
```

```python
analysis["Date"] = pd.to_datetime(analysis["Date"])
analysis["Day"] = analysis["Date"].dt.day_name()

analysis["Day"].value_counts()
```

```python
analysis["Month"] = analysis["Date"].dt.month_name()

analysis.groupby("Month")["Sales"].sum().sort_values(ascending=False)
```

```python
analysis[analysis["Sales"] > 500]
```

```python
analysis.sort_values("Sales", ascending=False).head(5)
```

```python
print("Total Revenue:", analysis["Sales"].sum())
print("Total Transactions:", analysis["Invoice ID"].nunique())
print("Average Basket Size:", analysis["Sales"].mean())
```

```python
analysis.columns
```

```python
branch_summary = analysis.groupby("Branch").agg(
    Total_Revenue=("Sales", "sum"),
    Total_Transactions=("Invoice ID", "count"),
    Average_Sales=("Sales", "mean")
).reset_index()

branch_summary
```

```python
analysis.groupby("Branch")["Sales"].sum().plot(kind="bar", title="Sales by Branch")
```

```python
analysis.groupby("Product line")["Sales"].sum().plot(kind="bar", title="Sales by Product Line")
```

```python
analysis["Payment"].value_counts().plot(kind="bar", title="Payment Methods")
```

insights
The analysis shows the total revenue and total number of transactions in the supermarket dataset.
The branch-wise analysis identifies differences in total sales and average basket size.
The product-line analysis shows which product lines generate higher sales.
The day-wise and month-wise analysis identifies the periods with higher transaction activity.
The analysis of payment methods shows the commonly used payment options among customers.
